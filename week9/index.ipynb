```python
import requests # นำเข้าโมดูล requests
url = "http://quotes.toscrape.com/" # กำหนดที่อยู่เว็บไซต์ (URL)

response = requests.get(url) # ส่งคำขอไปยัง URL และรับข้อมูลกลับมา

print("Status Code:", response.status_code) # แสดง HTTP Status Code

# แสดงเนื้อหา HTML ทั้งหมดของหน้าเว็บนั้นออกมาเป็นข้อความ (String)
# print(response.text)
print(type(response.text))
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(response.text,'html.parser')

tag = soup.h1
print("tag h1:\n",tag,"\n")

print("in tag hi:\n",tag.a,"\n")

print("attr tag a :\n",tag.a.attrs,"\n")

print("string in tag a:\n",tag.a.string)
```

```python
html = """ 
<div id="main" class="container">
    <h1 class="title">IT Shop</h1>
    <div class="product-list">
        <div class="item" id="p1">
            <h3>Laptop</h3>
            <p class="price">30,000 THB</p>
            <a href="/buy/p1">Buy</a>
        </div>
        
        <div class="item" id="p2">
            <h3>Mouse</h3>
            <p class="price">500 THB</p>
            <a href="/buy/p2">Buy</a>
        </div>
    </div>
</div> """
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(html,'html.parser')
title = soup.find("h1")
print(title)
```

```python
links = soup.find_all("a")

for link in links:
    print(link.get("href"))
```

```python
links = soup.find_all(["a","p"])

for link in links:
    print(link)
```

```python
itemId_1 = soup.find(id="p1")
print(itemId_1)
```

```python
price = soup.find_all(class_="price")

for i in price:
    print(i.text)
```

```python
product_list = soup.find(class_="product-list")

print(len(product_list.contents))

for i in product_list.children:
    if i.name:
        print(f"find child: {i.name}")
```

```python
price_tag = soup.find(class_="price")

parent_div = price_tag.parent

print(parent_div)
print(parent_div.get('id'))
```

```python
item = soup.find(class_="item")

print(item,"\n")
print(item.next_sibling.next_sibling)

item = item.h3
print("\n",item.next_sibling.next_sibling)
```
